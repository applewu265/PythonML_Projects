```python
import pandas as pd
import numpy
import matplotlib.pyplot as plt
import pathlib
```

```python
seeds = pd.read_csv("data/mens/MNCAATourneySeeds.csv")
tourney_compact = pd.read_csv("data/mens/MNCAATourneyCompactResults.csv")
season_detailed = pd.read_csv("data/mens/MRegularSeasonDetailedResults.csv")
```

structure the dataset like:\
row team1 team2 [features] result\
0   wteam lteam [...]      1\
1   lteam wteam [...].     0

```python
# create target variable y from tourney_compact
win_df = pd.DataFrame()
win_df["Team1"] = tourney_compact["WTeamID"]
win_df["Team2"] = tourney_compact["LTeamID"]
win_df["Result"] = 1

loss_df = pd.DataFrame()
loss_df["Team1"] = tourney_compact["LTeamID"]
loss_df["Team2"] = tourney_compact["WTeamID"]
loss_df["Result"] = 0

for col in tourney_compact.columns:
    if col in ["WTeamID", "LTeamID", "WLoc"]: continue
    if col[0]=="W":
        win_df[f"Team1_{col[1:]}"]=tourney_compact[col]
        loss_df[f"Team2_{col[1:]}"]=tourney_compact[col]
        continue
    if col[0]=="L":
        win_df[f"Team2_{col[1:]}"]=tourney_compact[col]
        loss_df[f"Team1_{col[1:]}"]=tourney_compact[col]
        continue
    else:
        win_df[col]=tourney_compact[col]
        loss_df[col]=tourney_compact[col]
        continue

# stack together
results_df = pd.concat([win_df, loss_df]).reset_index(drop=True)
```
